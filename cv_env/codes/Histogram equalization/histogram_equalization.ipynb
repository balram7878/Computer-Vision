```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("./images/x-ray.png", 0)

hist = np.bincount(img.ravel(), minlength=256)
equalized = cv2.equalizeHist(img)


# Results: images, histograms, CDFs
eq_hist = np.bincount(equalized.ravel(), minlength=256)


plt.figure(figsize=(11, 12))

plt.subplot(3, 2, 1); plt.imshow(img, cmap="gray", vmin=0, vmax=255); plt.title("Original"); plt.axis("off")
plt.subplot(3, 2, 2); plt.imshow(equalized, cmap="gray", vmin=0, vmax=255); plt.title("Equalized"); plt.axis("off")

plt.subplot(3, 2, 3); plt.bar(range(256), hist, width=1); plt.title("Original histogram"); plt.xlim(0, 255)
plt.subplot(3, 2, 4); plt.bar(range(256), eq_hist, width=1); plt.title("Equalized histogram"); plt.xlim(0, 255)


plt.tight_layout()
plt.show()
```
