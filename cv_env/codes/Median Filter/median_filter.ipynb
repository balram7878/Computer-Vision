```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("./dust.png", 0)

noisy = img.copy()
mask = np.random.rand(*img.shape)
noisy[mask < 0.03] = 0       # 3% pepper
noisy[mask > 0.97] = 255     # 3% salt

median = cv2.medianBlur(noisy, 3)
average = cv2.blur(noisy, (3, 3))

crop = lambda x: x[100:300, 100:300]
for i, (im, t) in enumerate([(noisy, "Salt & pepper"), (average, "Average 3x3"), (median, "Median 3x3")]):
    plt.subplot(1, 3, i + 1); plt.imshow(crop(im), cmap="gray", vmin=0, vmax=255); plt.title(t); plt.axis("off")
plt.show()
```
