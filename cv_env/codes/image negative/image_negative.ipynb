```python
import cv2

img = cv2.imread("./x-ray.png", 0)

negative = 255 - img

cv2.namedWindow("original image", cv2.WINDOW_NORMAL)
cv2.namedWindow("image negative", cv2.WINDOW_NORMAL)
cv2.resizeWindow("original image", 500, 500)
cv2.resizeWindow("image negative", 500, 500)

cv2.imshow("original image", img)
cv2.imshow("image negative", negative)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
