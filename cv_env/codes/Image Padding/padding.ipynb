```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("../../images/horse.png", 0)
if img is None:
    raise SystemExit("Image not found, check path")

p = 50   # pad 50 pixels so it's easy to see

zero      = np.pad(img, p, mode="constant", constant_values=0)
replicate = np.pad(img, p, mode="edge")
reflect   = np.pad(img, p, mode="reflect")

# OpenCV reference
ref = cv2.copyMakeBorder(img, p, p, p, p, cv2.BORDER_REPLICATE)
print("Replicate matches OpenCV:", np.array_equal(replicate, ref))

print("Original:", img.shape, "-> Padded:", zero.shape)

plt.figure(figsize=(12, 4))
for i, (name, im) in enumerate([("Zero", zero), ("Replicate", replicate), ("Reflect", reflect)]):
    plt.subplot(1, 3, i + 1)
    plt.imshow(im, cmap="gray", vmin=0, vmax=255)
    plt.title(name + " padding")
    plt.axis("off")
plt.show()
```
