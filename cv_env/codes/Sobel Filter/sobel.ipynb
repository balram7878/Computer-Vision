```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("../../images/lane2.png", 0)

gx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)   # x direction
gy = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)   # y direction
mag = cv2.convertScaleAbs(cv2.magnitude(gx, gy)) # clip to 0-255, uint8

imgs = [img, np.abs(gx), np.abs(gy), mag]
titles = ["Original", "|Gy|", "|Gx|", "Magnitude"]
plt.figure(figsize=(12, 6))                  # width, height in inches
for i, (im, t) in enumerate(zip(imgs, titles)):
    plt.subplot(1, 4, i + 1)                  # 2 rows, 2 columns
    plt.imshow(im, cmap="gray"); plt.title(t); plt.axis("off")
plt.tight_layout()
plt.show()
```
