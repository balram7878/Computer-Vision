```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("../../images/lane2.png", 0)
blur = cv2.GaussianBlur(img, (3, 3), 0)

lap = cv2.Laplacian(img, cv2.CV_64F)         # raw, has negative values
lap_blur = cv2.Laplacian(blur, cv2.CV_64F)   # smooth first

imgs = [img, np.abs(lap), np.abs(lap_blur)]
titles = ["Original", "Laplacian", "Blur then Laplacian"]
plt.figure(figsize=(15, 6))
for i, (im, t) in enumerate(zip(imgs, titles)):
    plt.subplot(1, 3, i + 1); plt.imshow(im, cmap="gray", vmax=np.abs(lap).max() / 3); plt.title(t); plt.axis("off")
plt.tight_layout(); plt.show()

print("Sharpness score (variance):", lap.var())
```
