```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("../../images/house.png", 0)

lap = cv2.Laplacian(img, cv2.CV_64F)
sharp = np.clip(img - lap, 0, 255).astype(np.uint8)

ref = cv2.filter2D(img, -1, np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]]))
print("Max difference vs one-kernel:", np.abs(sharp.astype(int) - ref).max())

plt.figure(figsize=(14, 6))
for i, (im, t) in enumerate([(img, "Original"), (sharp, "Laplacian sharpened")]):
    plt.subplot(1, 2, i + 1); plt.imshow(im, cmap="gray", vmin=0, vmax=255); plt.title(t); plt.axis("off")
plt.tight_layout(); plt.show()
```
