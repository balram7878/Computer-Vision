```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("./images/x-ray.png", 0)
low, high = 80, 180

manual = img.copy()
manual[manual < low] = low
manual[manual > high] = high

ref = np.clip(img, low, high)
print("Match:", np.array_equal(manual, ref))

plt.subplot(1,2,1); plt.imshow(img, cmap="gray", vmin=0, vmax=255); plt.title("Original")
plt.subplot(1,2,2); plt.imshow(manual, cmap="gray", vmin=0, vmax=255); plt.title(f"Clipped [{low}, {high}]")
plt.show()

plt.figure(figsize=(10,4))
plt.subplot(1,2,1); plt.hist(img.ravel(), bins=256, range=(0,255)); plt.title("Original histogram")
plt.subplot(1,2,2); plt.hist(manual.ravel(), bins=256, range=(0,255)); plt.title("Clipped histogram")
plt.show()
```
