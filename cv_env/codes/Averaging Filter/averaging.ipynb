```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("../../images/wolf.png", 0)

kernel = np.ones((4, 4)) / 9
print(kernel)
avg = cv2.filter2D(img, -1, kernel)     # shows the kernel explicitly
ref = cv2.blur(img, (3, 3))             # one-line shortcut
print("Max difference:", np.abs(avg.astype(int) - ref).max())

plt.subplot(1, 2, 1); plt.imshow(img, cmap="gray", vmin=0, vmax=255); plt.title("Original")
plt.subplot(1, 2, 2); plt.imshow(avg, cmap="gray", vmin=0, vmax=255); plt.title("3x3 average")
plt.show()
```
