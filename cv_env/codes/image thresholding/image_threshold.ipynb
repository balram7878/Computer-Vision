```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("./nuclei.png", 0)
T = 210

manual = ((img > T) * 255).astype(np.uint8)       # your own implementation
_, ref = cv2.threshold(img, T, 255, cv2.THRESH_BINARY)  # OpenCV reference

print("Match:", np.array_equal(manual, ref))

plt.subplot(1,2,1); plt.imshow(img, cmap="gray", vmin=0, vmax=255); plt.title("Original")
plt.subplot(1,2,2); plt.imshow(manual, cmap="gray", vmin=0, vmax=255); plt.title(f"Threshold T={T}")
plt.show()
```
